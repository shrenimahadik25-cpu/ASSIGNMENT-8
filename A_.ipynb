```python
# A* Search Algorithm using Graph

import heapq

# Define the graph with edge costs
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [('G', 4)],
    'E': [('G', 2)],
    'F': [('G', 2)],
    'G': []
}

# Heuristic values (estimated distance to goal G)
heuristic = {
    'A': 7,
    'B': 6,
    'C': 4,
    'D': 4,
    'E': 2,
    'F': 2,
    'G': 0
}

def astar(start, goal):
    # Priority queue stores (f_cost, current_node)
    open_list = [(heuristic[start], start)]

    # Store actual cost from start
    g_cost = {start: 0}

    # Store parent nodes to reconstruct the path
    parent = {start: None}

    while open_list:
        f, current = heapq.heappop(open_list)

        # Goal reached
        if current == goal:
            path = []
            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()
            return path, g_cost[goal]

        # Explore neighboring nodes
        for neighbor, cost in graph[current]:
            new_cost = g_cost[current] + cost

            if neighbor not in g_cost or new_cost < g_cost[neighbor]:
                g_cost[neighbor] = new_cost
                parent[neighbor] = current

                # f(n) = g(n) + h(n)
                f_cost = new_cost + heuristic[neighbor]

                heapq.heappush(open_list, (f_cost, neighbor))

    return None, float('inf')


# Execute the algorithm
start_node = 'A'
goal_node = 'G'

path, cost = astar(start_node, goal_node)

# Display the result
print("A* Search Algorithm")
print("--------------------")
print("Start Node:", start_node)
print("Goal Node:", goal_node)

if path:
    print("Shortest Path:", " -> ".join(path))
    print("Total Path Cost:", cost)
else:
    print("No path found!")
```
